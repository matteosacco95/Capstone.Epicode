```python
import pandas as pd

def filter_uk_major_cities(input_file, output_file):
    # List of major UK cities to keep
    major_uk_cities = [
        "London", "Birmingham", "Manchester", "Leeds", "Glasgow", 
        "Sheffield", "Bradford", "Liverpool", "Edinburgh", "Bristol",
        "Cardiff", "Leicester", "Coventry", "Nottingham", "Newcastle upon Tyne",
        "Brighton", "Hull", "Plymouth", "Stoke-on-Trent", "Wolverhampton",
        "Derby", "Southampton", "Portsmouth", "Aberdeen", "Northampton"
    ]

 # Target Country String
    target_country = "United Kingdom of Great Britain and Northern Ireland"

    try:
        # Load the dataset
        df = pd.read_csv(input_file)
        
        # Check for required columns
        required_cols = ['City', 'Country']
        missing_cols = [col for col in required_cols if col not in df.columns]
        
        if missing_cols:
            print(f"Error: Missing required columns: {missing_cols}")
            print(f"Available columns: {df.columns.tolist()}")
            return

        # Step 1: Filter by Country
        # We normalize the country string to handle potential case differences or extra spaces
        # while ensuring the core meaning matches the target.
        # Note: Using .str.strip() and .str.title() might alter the exact string match if the 
        # source data has specific capitalization. Here we do a case-insensitive strip check.
        df['Country_Clean'] = df['Country'].astype(str).str.strip()
        
        # Filter for the specific country string (case-insensitive match for safety)
        uk_mask = df['Country_Clean'].str.lower() == target_country.lower()
        df_uk = df[uk_mask].copy()

        if df_uk.empty:
            print(f"Warning: No rows found with Country = '{target_country}'.")
            print("Sample unique countries in your data (first 5):")
            print(df['Country'].unique()[:5])
            return

        # Step 2: Filter by Major Cities
        # Normalize city names for comparison
        df_uk['City_Clean'] = df_uk['City'].astype(str).str.strip().str.title()
        
        final_df = df_uk[df_uk['City_Clean'].isin(major_uk_cities)]

        if final_df.empty:
            print("Warning: No matching major UK cities found within the UK dataset.")
            print("Sample unique cities in the UK subset (first 10):")
            print(df_uk['City'].unique()[:10])
        else:
            # Drop the temporary helper columns before saving
            cols_to_drop = [c for c in final_df.columns if c.endswith('_Clean')]
            final_df = final_df.drop(columns=cols_to_drop)

            # Save the filtered data
            final_df.to_csv(output_file, index=False)
            print(f"Success!")
            print(f"Input file: {input_file}")
            print(f"Output file: {output_file}")
            print(f"Total rows in input: {len(df)}")
            print(f"Rows after Country filter: {len(df_uk)}")
            print(f"Final rows (Major UK Cities): {len(final_df)}")
            
    except FileNotFoundError:
        print(f"Error: Input file not found at '{input_file}'.")
        print("Please ensure 'air pollution dataset.csv' is in the same folder as this script.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Run the function
filter_uk_major_cities('air pollution dataset.csv', 'uk_major_cities_pollution.csv')
```
