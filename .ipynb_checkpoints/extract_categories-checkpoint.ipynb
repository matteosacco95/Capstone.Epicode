```python
import pandas as pd
import os
import re

def extract_categories_to_csvs(file_path, output_dir="categories"):
    """
    Reads the DFT road casualty data guide Excel file and splits 
    categorical fields into individual CSV files.
    """
    
    # Create output directory if it doesn't exist
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        print(f"Created directory: {output_dir}")

    try:
        # Read the Excel file
        # The file appears to be a single sheet with a dictionary structure
        df = pd.read_excel(file_path)
        
        print(f"Loaded file: {file_path}")
        print(f"Total rows in dictionary: {len(df)}")
        
        # Identify columns
        # Based on the file content, columns are: table, field name, code/format, label, note
        # We need to ensure we handle potential whitespace in column names
        df.columns = df.columns.str.strip()
        
        required_cols = ['field name', 'code/format', 'label']
        if not all(col in df.columns for col in required_cols):
            print("Error: Expected columns not found. Available columns:", df.columns.tolist())
            return

        # Filter rows that actually contain category data
        # Categories usually have a numeric or short string code and a text label
        # We exclude rows where 'code/format' is empty or just a description (like ranges)
        # We also exclude rows where 'label' is empty
        
        # Clean the dataframe: drop rows with missing codes or labels
        clean_df = df.dropna(subset=['code/format', 'label'])
        
        # Group by 'field name' to process each category separately
        grouped = clean_df.groupby('field name')
        
        count = 0
        for field_name, group in grouped:
            # Skip fields that look like they are just ranges or descriptions without discrete codes
            # e.g., "1 to 9999" or "See latest ONS code list"
            sample_code = group['code/format'].iloc[0]
            if isinstance(sample_code, str) and ('to' in sample_code.lower() or 'see' in sample_code.lower()):
                continue
            
            # Create a clean dataframe for this category
            # Select only relevant columns and rename for clarity
            cat_df = group[['code/format', 'label']].copy()
            cat_df.columns = ['code', 'label']
            
            # Add a 'note' column if it exists and has data
            if 'note' in group.columns:
                notes = group['note'].dropna()
                if not notes.empty:
                    cat_df['note'] = group['note']
            
            # Clean the filename: replace spaces/special chars with underscores
            safe_filename = re.sub(r'[^\w\-_\. ]', '_', field_name)
            safe_filename = safe_filename.replace(' ', '_').lower()
            filename = f"{safe_filename}.csv"
            filepath = os.path.join(output_dir, filename)
            
            # Save to CSV
            cat_df.to_csv(filepath, index=False)
            count += 1
            print(f"Saved: {filename} ({len(cat_df)} rows)")

        print(f"\nSuccessfully extracted {count} category files to '{output_dir}' folder.")

    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    # Update this path to match your file location
    input_file = "dft-road-casualty-statistics-road-safety-open-dataset-data-guide-2024.xlsx"
    
    # Run the extraction
    extract_categories_to_csvs(input_file)
```
