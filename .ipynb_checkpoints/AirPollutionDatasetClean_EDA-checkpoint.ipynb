```python
import pandas as pd
import numpy as np
import os

def create_star_schema_separate_dims(input_file, output_dir="star_schema_output"):    
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    print("="*60)
    print("STAR SCHEMA CREATION (SEPARATE DIMENSIONS)")
    print("="*60)
    
    # 1. LOAD DATA & SANITIZE
    print("\nStep 1: Loading dataset and sanitizing column names...")
    try:
        df = pd.read_csv(input_file)
        
        # CRITICAL: Strip whitespace from ALL column names to prevent KeyError
        df.columns = df.columns.str.strip()
        
        original_row_count = len(df)
        print(f"Original rows: {original_row_count}")
        print(f"Columns found: {list(df.columns)}")
    except Exception as e:
        print(f"Error loading file: {e}")
        return

    # 2. DROP 'newcountry' COLUMN
    print("\nStep 2: Dropping 'newcountry' column...")
    cols_to_drop = []
    for col in df.columns:
        if col.lower() == 'newcountry':
            cols_to_drop.append(col)
            print(f"  [FOUND] Dropping: '{col}'")
    
    if cols_to_drop:
        df.drop(columns=cols_to_drop, inplace=True)
        print(f"  [OK] Dropped {len(cols_to_drop)} column(s).")
    else:
        print("  [INFO] No 'newcountry' column found.")

    # 3. HANDLE MISSING VALUES
    print("\nStep 3: Handling missing values...")
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    cat_cols = df.select_dtypes(include=['object']).columns

    for col in numeric_cols:
        if df[col].isnull().any():
            median_val = df[col].median()
            df.loc[df[col].isnull(), col] = median_val
            
    for col in cat_cols:
        if df[col].isnull().any():
            df.loc[df[col].isnull(), col] = 'Unknown'

    # 4. IDENTIFY COLUMNS
    print("\nStep 4: Identifying columns...")
    
    # Identify City (Primary Key Candidate)
    city_col = None
    for col in df.columns:
        if 'city' in col.lower() or 'location' in col.lower():
            city_col = col
            break
    
    # Identify Country (for separate dimension)
    country_col = None
    for col in df.columns:
        if 'country' in col.lower() and 'new' not in col.lower():
            country_col = col
            break

    # Identify Category/Index columns for SEPARATE dimensions
    # Keywords: 'category', 'index', 'level', 'class', 'quality', 'status'
    category_keywords = ['category', 'index', 'level', 'class', 'quality', 'status']
    category_cols = []
    
    for col in df.columns:
        if col in [city_col, country_col]:
            continue
        
        col_lower = col.lower()
        if any(k in col_lower for k in category_keywords):
            category_cols.append(col)
    
    print(f"  Detected City Column: {city_col}")
    print(f"  Detected Country Column: {country_col}")
    print(f"  Detected Category Columns (will get SEPARATE tables): {category_cols}")

    # 5. BUILD DIMENSION TABLES
    print("\nStep 5: Creating Dimension Tables...")
    dimension_tables = {}
    fact_table = df.copy()
    
    # 5A: Country Dimension (Separate)
    if country_col:
        print(f"  Creating Country Dimension: '{country_col}'...")
        unique_vals = fact_table[[country_col]].drop_duplicates().reset_index(drop=True)
        dim_country = unique_vals.copy()
        dim_country.insert(0, "country_id", range(1, len(dim_country) + 1))
        dim_country.to_csv(os.path.join(output_dir, "dim_country.csv"), index=False, encoding='utf-8')
        dimension_tables['country'] = dim_country
        
        mapping = dict(zip(dim_country[country_col], dim_country['country_id']))
        fact_table['country_id'] = fact_table[country_col].map(mapping)
        fact_table.drop(columns=[country_col], inplace=True)
        print(f"  [OK] Created dim_country.csv ({len(dim_country)} entries)")

    # 5B: City Dimension (Primary Key Source)
    if city_col:
        print(f"  Creating City Dimension: '{city_col}'...")
        unique_vals = fact_table[[city_col]].drop_duplicates().reset_index(drop=True)
        dim_city = unique_vals.copy()
        dim_city.insert(0, "city_id", range(1, len(dim_city) + 1))
        dim_city.to_csv(os.path.join(output_dir, "dim_city.csv"), index=False, encoding='utf-8')
        dimension_tables['city'] = dim_city
        
        mapping = dict(zip(dim_city[city_col], dim_city['city_id']))
        fact_table['city_id'] = fact_table[city_col].map(mapping)
        fact_table.drop(columns=[city_col], inplace=True)
        print(f"  [OK] Created dim_city.csv ({len(dim_city)} entries)")
    else:
        print("  [WARN] No City column found. Cannot use city_id as primary key.")

    # 5C: SEPARATE Category Dimensions (ONE table PER column)
    if category_cols:
        print(f"  Creating SEPARATE dimension tables for {len(category_cols)} category columns...")
        
        for col in category_cols:
            if col not in fact_table.columns:
                print(f"    [SKIP] Column '{col}' not found. Skipping.")
                continue
            
            print(f"    Processing '{col}'...")
            
            # Get unique values for THIS specific column
            unique_vals = fact_table[[col]].drop_duplicates().reset_index(drop=True)
            
            # Create dimension table
            dim_table = unique_vals.copy()
            # Create ID column name based on the source column (e.g., aqi_category_id)
            id_col_name = f"{col}_id"
            dim_table.insert(0, id_col_name, range(1, len(dim_table) + 1))
            
            # Save dimension table (sanitize filename)
            safe_filename = f"dim_{col}.csv".replace(" ", "_").replace("-", "_").lower()
            dim_table.to_csv(os.path.join(output_dir, safe_filename), index=False, encoding='utf-8')
            dimension_tables[col] = dim_table
            
            # Map values to IDs in Fact table
            mapping = dict(zip(dim_table[col], dim_table[id_col_name]))
            fact_table[id_col_name] = fact_table[col].map(mapping)
            
            # Handle unmapped values
            if fact_table[id_col_name].isnull().any():
                print(f"      [WARN] Unmapped values in '{col}'. Filling with 0.")
                fact_table[id_col_name] = fact_table[id_col_name].fillna(0)
            
            # Drop original text column
            fact_table.drop(columns=[col], inplace=True)
            print(f"      [OK] Created {safe_filename} ({len(dim_table)} entries). Replaced '{col}' with '{id_col_name}'.")
        
        print(f"  [OK] Created {len(category_cols)} separate category dimension tables.")

    # 6. FINALIZE FACT TABLE (NO synthetic fact_id)
    print("\nStep 6: Finalizing Fact Table (Using city_id as Primary Key)...")
    
    # Reorder columns: city_id first, then other IDs, then measures
    id_cols = [col for col in fact_table.columns if col.endswith('_id')]
    non_id_cols = [col for col in fact_table.columns if not col.endswith('_id')]
    
    # Ensure city_id is first
    if 'city_id' in id_cols:
        id_cols.remove('city_id')
        final_order = ['city_id'] + id_cols + non_id_cols
    else:
        # Fallback if city_id missing
        final_order = id_cols + non_id_cols

    fact_table = fact_table[final_order]

    fact_table.to_csv(os.path.join(output_dir, "fact_measurements.csv"), index=False, encoding='utf-8')
    print(f"  [OK] Created fact_measurements.csv ({len(fact_table)} records)")
    print(f"  [OK] Columns: {list(fact_table.columns)}")
    print(f"  [OK] Primary Key Strategy: 'city_id' (Natural Key)")

    # 7. VERIFICATION
    print("\nStep 7: Verification...")
    final_fact_count = len(fact_table)
    if final_fact_count != original_row_count:
        print(f"  [WARN] Row count mismatch! Original: {original_row_count}, Final: {final_fact_count}")
    else:
        print(f"  [OK] Row count preserved: {final_fact_count} rows")

    # 8. SAVE SCHEMA DOC
    schema_doc = os.path.join(output_dir, "schema_structure.txt")
    with open(schema_doc, 'w', encoding='utf-8') as f:
        f.write("STAR SCHEMA STRUCTURE (SEPARATE DIMENSIONS)\n")
        f.write("="*60 + "\n\n")
        f.write(f"Original Rows: {original_row_count}\n")
        f.write(f"Final Fact Rows: {final_fact_count}\n\n")
        f.write("USER REQUESTS IMPLEMENTED:\n")
        f.write("  1. Dropped 'newcountry' column.\n")
        f.write("  2. Created SEPARATE dimension tables for EACH category column.\n")
        f.write("  3. Used 'city_id' as Primary Key (No synthetic fact_id).\n\n")
        f.write("FACT TABLE: fact_measurements\n")
        f.write(f"  Primary Key: city_id\n")
        f.write(f"  Foreign Keys: {[c for c in id_cols if c != 'city_id']}\n")
        f.write(f"  Measures: {non_id_cols}\n\n")
        f.write("DIMENSION TABLES:\n")
        for dim_name, df_dim in dimension_tables.items():
            f.write(f"  - dim_{dim_name}.csv\n")
            if '_' in dim_name:
                pk_name = f"{dim_name.split('_')[-1]}_id"
            else:
                pk_name = f"{dim_name}_id"
            f.write(f"    Primary Key: {pk_name}\n")
            f.write(f"    Unique Values: {len(df_dim)}\n")

    print(f"\n{'='*60}")
    print("FILES SAVED TO:", os.path.abspath(output_dir))
    print("="*60)
    
    return dimension_tables, fact_table

if __name__ == "__main__":
    input_file = "uk_major_cities_pollution.csv"
    output_folder = "star_schema_output"
    
    if os.path.exists(input_file):
        create_star_schema_separate_dims(input_file, output_folder)
    else:
        print(f"Error: File '{input_file}' not found.")
```
