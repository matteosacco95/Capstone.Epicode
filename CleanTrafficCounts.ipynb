```python
import pandas as pd

# Define the input and output filenames
input_file = 'dft_traffic_counts_raw_counts.csv'
output_file = 'dft_traffic_counts_raw_counts_reduced.csv'

# List of columns to keep (Added 'year' to the list)
columns_to_keep = [
    'count_date',
    'hour',
    'pedal_cycles',
    'two_wheeled_motor_vehicles',
    'cars_and_taxis',
    'buses_and_coaches',
    'year'  # Added this column
]

try:
    # Read the CSV file
    df = pd.read_csv(input_file)
    
    # Check if all required columns exist
    missing_cols = [col for col in columns_to_keep if col not in df.columns]
    
    if missing_cols:
        print(f"Error: The following columns were not found in '{input_file}':")
        print(missing_cols)
        print("\nAvailable columns in the file:")
        print(df.columns.tolist())
    else:
        # Ensure 'year' is numeric to allow comparison
        # This handles cases where the column might be read as string
        df['year'] = pd.to_numeric(df['year'], errors='coerce')
        
        # Filter for years between 2020 and 2024 (inclusive)
        filtered_df = df[(df['year'] >= 2020) & (df['year'] <= 2024)]
        
        # Check if any rows remain after filtering
        if filtered_df.empty:
            print("Warning: No data found for years 2020-2024. The output file will be empty.")
        else:
            # Select only the desired columns
            reduced_df = filtered_df[columns_to_keep]
            
            # Save to a new CSV file
            reduced_df.to_csv(output_file, index=False)
            print(f"Success! Created '{output_file}' with {len(reduced_df)} rows (filtered for 2020-2024) and {len(columns_to_keep)} columns.")
            print("First few rows of the reduced dataset:")
            print(reduced_df.head())

except FileNotFoundError:
    print(f"Error: The file '{input_file}' was not found.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
