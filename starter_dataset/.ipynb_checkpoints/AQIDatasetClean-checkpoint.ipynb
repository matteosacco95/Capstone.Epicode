```python
"""
transform.py
------------
Transforms UK-Air hourly measurement CSVs (2020–2024) into a
PowerBI-friendly star schema:

  location.csv   — one row per monitoring site (with location_id PK)
  data.csv       — one row per measurement per site per hour
                   (with location_id FK + year column)

Input files expected in the same directory as this script:
    2020.csv, 2021.csv, 2022.csv, 2023.csv, 2024.csv

Output files written to the same directory:
    location.csv, data.csv
"""

import csv
import os
import re
from pathlib import Path

# ── CONFIG ────────────────────────────────────────────────────────────────────

INPUT_DIR = Path(__file__).parent          # folder containing this script
OUTPUT_DIR = Path(__file__).parent
YEARS = [2020, 2021, 2022, 2023, 2024]

# Row indices (0-based) in the 10-row metadata header
META_ROWS = {
    "site_name":      3,
    "latitude":       4,
    "longitude":      5,
    "site_type":      6,
    "zone":           7,
    "agglomeration":  8,
    "local_authority":9,
}

# Row index of the measurement/column-header row (0-based)
COLUMN_HEADER_ROW = 10

# ── HELPERS ───────────────────────────────────────────────────────────────────

def strip_unit(value: str) -> str:
    """Remove unit suffix from a measurement value, keep numeric part."""
    if not value or value.strip().lower() in ("", "no data"):
        return ""
    # Values look like "44.9 V ugm-3 (TEOM FDMS)" — keep only the number
    match = re.match(r"^\s*([0-9]+\.?[0-9]*(?:[eE][+-]?[0-9]+)?)", value.strip())
    return match.group(1) if match else ""

def parse_metadata(rows: list[list[str]]) -> dict[int, dict]:
    """
    Given all raw rows of a file, return a dict mapping
    column_index -> site metadata dict.
    Only columns that have a non-empty site_name are included.
    """
    sites = {}
    site_name_row = rows[META_ROWS["site_name"]]
    for col_idx, cell in enumerate(site_name_row):
        name = cell.strip()
        if name and name.lower() != "site name":
            meta = {"site_name": name}
            for field, row_idx in META_ROWS.items():
                if field == "site_name":
                    continue
                val = rows[row_idx][col_idx] if col_idx < len(rows[row_idx]) else ""
                meta[field] = val.strip()
            sites[col_idx] = meta
    return sites

def build_column_map(header_row: list[str], site_col_indices: list[int]) -> list[dict]:
    """
    Map each site's measurement columns to their measurement type.
    Returns a list of dicts: {site_col, measurement_type, status_col}
    """
    mapping = []
    # Date and Time are always cols 0 and 1
    # For each site, columns come in pairs: value, status
    for i, col_idx in enumerate(site_col_indices):
        mtype = header_row[col_idx].strip() if col_idx < len(header_row) else "Unknown"
        status_col = col_idx + 1 if (col_idx + 1) < len(header_row) else None
        mapping.append({
            "site_col": col_idx,
            "measurement_type": mtype,
            "status_col": status_col,
        })
    return mapping

# ── MAIN ──────────────────────────────────────────────────────────────────────

def main():
    all_locations: dict[str, dict] = {}   # site_name -> metadata (deduped)
    location_id_map: dict[str, int] = {}  # site_name -> location_id
    next_location_id = 1

    all_data_rows: list[dict] = []

    for year in YEARS:
        fpath = INPUT_DIR / f"{year}.csv"
        if not fpath.exists():
            print(f"  [SKIP] {fpath} not found")
            continue

        print(f"Processing {year}.csv ...")

        with open(fpath, newline="", encoding="utf-8-sig") as f:
            rows = list(csv.reader(f))

        # ── Parse site metadata ───────────────────────────────────────────────
        sites_meta = parse_metadata(rows)  # col_idx -> meta dict
        site_col_indices = sorted(sites_meta.keys())

        # Register new locations
        for col_idx in site_col_indices:
            meta = sites_meta[col_idx]
            name = meta["site_name"]
            if name not in location_id_map:
                location_id_map[name] = next_location_id
                all_locations[name] = meta
                next_location_id += 1

        # ── Parse column header row ───────────────────────────────────────────
        col_header = rows[COLUMN_HEADER_ROW]
        col_map = build_column_map(col_header, site_col_indices)

        # ── Parse data rows ───────────────────────────────────────────────────
        data_rows = rows[COLUMN_HEADER_ROW + 1:]
        for data_row in data_rows:
            if not data_row or not data_row[0].strip():
                continue
            date = data_row[0].strip()
            time = data_row[1].strip() if len(data_row) > 1 else ""

            for i, col_idx in enumerate(site_col_indices):
                site_name = sites_meta[col_idx]["site_name"]
                location_id = location_id_map[site_name]
                mtype = col_map[i]["measurement_type"]
                status_col = col_map[i]["status_col"]

                raw_value = data_row[col_idx].strip() if col_idx < len(data_row) else ""
                raw_status = data_row[status_col].strip() if (status_col and status_col < len(data_row)) else ""

                value = strip_unit(raw_value)
                # Extract just the status letter (first character: V, P, N, S)
                status = raw_status[0] if raw_status else ""

                all_data_rows.append({
                    "location_id": location_id,
                    "year": year,
                    "date": date,
                    "time": time,
                    "measurement_type": mtype,
                    "value": value,
                    "status": status,
                })

        print(f"  → {len(site_col_indices)} sites, {len(data_rows)} time rows")

    # ── Write location.csv ────────────────────────────────────────────────────
    loc_path = OUTPUT_DIR / "location.csv"
    loc_fields = ["location_id", "site_name", "latitude", "longitude",
                  "site_type", "zone", "agglomeration", "local_authority"]
    with open(loc_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=loc_fields)
        writer.writeheader()
        for name, meta in all_locations.items():
            row = {"location_id": location_id_map[name]}
            row.update(meta)
            writer.writerow(row)
    print(f"\n✓ location.csv → {len(all_locations)} sites")

    # ── Write data.csv ────────────────────────────────────────────────────────
    data_path = OUTPUT_DIR / "data.csv"
    data_fields = ["location_id", "year", "date", "time",
                   "measurement_type", "value", "status"]
    with open(data_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=data_fields)
        writer.writeheader()
        writer.writerows(all_data_rows)
    print(f"✓ data.csv     → {len(all_data_rows):,} measurement rows")
    print("\nDone.")


if __name__ == "__main__":
    main()
```
